## 📚 Explication 1 — Représenter et importer une table

### 🎯 Objectif de ce bloc
> Importer une table depuis un fichier CSV sous forme de liste de dictionnaires.

Une table, c'est comme un tableur : des **colonnes** (les descripteurs, ex : nom, classe, note) et des **lignes** (une ligne = un enregistrement). En Python, la façon la plus lisible de représenter une table est une **liste de dictionnaires**, un dictionnaire par ligne, où les clés sont les noms de colonnes.

```python
table = [
    {"nom": "Léa", "classe": "TG1", "note": "15"},
    {"nom": "Tom", "classe": "TG2", "note": "8"},
    {"nom": "Zoé", "classe": "TG1", "note": "12"}
]
print(table[0]["nom"])     # affiche "Léa"
print(table[1]["classe"])  # affiche "TG2"
```

Pour importer un vrai fichier CSV, on utilise le module `csv` avec `DictReader`, qui construit automatiquement cette liste de dictionnaires à partir des en-têtes de colonnes :

```python
import csv

table = []
with open("eleves.csv", newline="", encoding="utf-8") as fichier:
    lecteur = csv.DictReader(fichier)
    for ligne in lecteur:
        table.append(ligne)

print(table[0])   # {'nom': 'Léa', 'classe': 'TG1', 'note': '15'}
```



**Point de vigilance essentiel** : toutes les valeurs importées d'un CSV sont des **chaînes de caractères**, même si elles ressemblent à des nombres. `table[0]["note"]` vaut la chaîne `"15"`, pas l'entier `15`. Il faut convertir explicitement avant tout calcul ou tri numérique :
```python
note_numerique = int(table[0]["note"])
```

## 📚 Explication 2 — Rechercher des lignes selon un critère

### 🎯 Objectif de ce bloc
> Rechercher les lignes d'une table vérifiant un ou plusieurs critères combinés.

Rechercher dans une table, c'est parcourir la liste et ne garder que les lignes qui vérifient une condition — exactement comme une recherche séquentielle dans une liste simple, mais en testant une colonne précise.

```python
table = [
    {"nom": "Léa", "classe": "TG1", "note": 15},
    {"nom": "Tom", "classe": "TG2", "note": 8},
    {"nom": "Zoé", "classe": "TG1", "note": 12}
]

resultat = [ligne for ligne in table if ligne["note"] >= 10]
print(resultat)   # lignes de Léa et Zoé
```

On peut combiner plusieurs critères avec `and`/`or` :
```python
resultat = [ligne for ligne in table if ligne["classe"] == "TG1" and ligne["note"] >= 12]
print(resultat)   # seulement la ligne de Zoé
```

**Point de vigilance** : bien placer les parenthèses quand on mélange `and` et `or`, sinon la condition ne teste pas ce qu'on croit.

Vérification : que renverrait `[ligne for ligne in table if ligne["classe"] == "TG1" or ligne["note"] >= 12]` sur cette table ?

## 📚 Explication 3 — Trier une table

### 🎯 Objectif de ce bloc
> Trier une table suivant une colonne donnée.

On trie une table avec `sorted(table, key=...)`, où `key` indique quelle colonne utiliser pour comparer les lignes entre elles.

```python
table = [
    {"nom": "Léa", "note": 15},
    {"nom": "Tom", "note": 8},
    {"nom": "Zoé", "note": 12}
]

table_triee = sorted(table, key=lambda ligne: ligne["note"])
print(table_triee)   # triée par note croissante : Tom, Zoé, Léa

table_triee_desc = sorted(table, key=lambda ligne: ligne["note"], reverse=True)
print(table_triee_desc)   # triée par note décroissante : Léa, Zoé, Tom
```

**Point de vigilance** : `sorted(table, key=lambda ligne: ligne["note"])` ne fonctionne correctement que si `ligne["note"]` est bien un nombre — si c'est encore une chaîne issue d'un CSV, le tri se fera dans l'ordre alphabétique ("10" viendrait avant "9" !) et non dans l'ordre numérique.

Vérification : pourquoi trier des chaînes `"9"`, `"10"`, `"2"` donne-t-il l'ordre `"10"`, `"2"`, `"9"` et pas `"2"`, `"9"`, `"10"` ?

## 📚 Explication 4 — Fusionner deux tables

### 🎯 Objectif de ce bloc
> Construire une nouvelle table en combinant les données de deux tables partageant une colonne commune.

Fusionner deux tables, c'est combiner leurs colonnes pour les lignes qui partagent une même valeur dans une colonne commune (souvent appelée **clé**), comme le nom.


```python
notes = [
    {"nom": "Léa", "note": 15},
    {"nom": "Tom", "note": 8}
]
classes = [
    {"nom": "Léa", "classe": "TG1"},
    {"nom": "Tom", "classe": "TG2"}
]

fusion = []
for ligne_notes in notes:
    for ligne_classes in classes:
        if ligne_notes["nom"] == ligne_classes["nom"]:
            nouvelle_ligne = {"nom": ligne_notes["nom"], "note": ligne_notes["note"], "classe": ligne_classes["classe"]}
            fusion.append(nouvelle_ligne)

print(fusion)
```



**Point de vigilance** : si la colonne commune contient des doublons (deux lignes avec le même nom), la fusion peut créer plusieurs lignes fusionnées pour un même nom — il faut s'assurer que la colonne utilisée comme clé est bien unique dans chaque table.

Vérification : que se passerait-il si `notes` contenait deux lignes avec `"nom": "Léa"` ?

## ✅ Bilan de la notion
- Une table se représente en Python comme une liste de dictionnaires, une ligne = un dictionnaire, les colonnes = les clés.
- Un CSV importé avec `csv.DictReader` donne des chaînes de caractères pour toutes les valeurs — il faut convertir avant tout calcul ou tri numérique.
- Rechercher dans une table, c'est filtrer les lignes vérifiant un ou plusieurs critères combinés avec `and`/`or`.
- Trier une table se fait avec `sorted(table, key=lambda ligne: ligne["colonne"])`.
- Fusionner deux tables combine leurs colonnes pour les lignes partageant une valeur commune dans une colonne clé — attention aux doublons et aux lignes sans correspondance.
