# TP modularité La banque

La Banque des Canard (BdC) est une institution Implémentation
nouvellement implantée en Gers. Outre un compte
courant classique, elle propose également à ses clients, un
compte rémunéré à un taux qui peut éventuellement varier.
À une date fixée, la banque calcule les intérêts et crédite les
comptes rémunérés des clients. Elle vous demande de créer
un logiciel pour administrer correctement sa clientèle.


## Cahier des charges

 La Banque possède la liste de tous ses clients. Elle peut
enregistrer un nouveau client. Dans ce cas elle lui
ouvre automatiquement un Compte non rémunéré et un
Compte rémunéré. Enfin elle peut également appliquer
un taux de rémunération de 3% à tous les comptes
rémunérés

 Un Client est identifié par son nom et son prénom. On
considèrera qu’il ne peut y avoir d’homonyme. Il
possède deux Comptes (rémunéré et non rémunéré). Il
peut alimenter ou utiliser l’argent sur n’importe lequel
de ses comptes.


Un Compte est identifié par un numéro unique (fourni
par la Banque). Il possède une somme d’argent,
initialisée à 0 à l’ouverture. Ce compte peut être
rémunéré ou non. Ce compte peut être crédité ou
débité. En cas de débit supérieur à la somme présente
sur le compte, le retrait est refusé.

## Programmation modulaire

Un projet important implique plusieurs programmeurs.
On le découpe en plusieurs objets.


Chaque personne n’a pas besoin de connaître
l’implémentation de chaque classe. Ce dernier lui fournit
par contre une interface qui définit ce que peut faire la
classe. 

La classe garantit ce que va faire l’objet (mais pas
comment le programmeur va s’en servir).
