### Étape 1 — C'est quoi un objet ?

Imagine que tu veux décrire une voiture dans ton programme. Cette voiture a :
- des **caractéristiques** (une marque, une couleur, une immatriculation)

- les **attributs** = les caractéristiques
- les **méthodes** = les actions

## La classe : le moule

 
Une **classe**, c'est comme un moule à gaufres. Le moule ne mange pas, il ne se mange pas — mais il permet de fabriquer autant de gaufres qu'on veut, toutes avec la même forme.     
  
 Une classe détaille le squelette d’un objet. Elle définit:  

- les attributs: il s’agit des caractéristiques spécifiques de l’objet,
- les méthodes: il s’agit des fonctionnalités que peut réaliser l’objet. 



``` 
CLASSE Voiture
  ├── attribut : marque
  ├── attribut : couleur
  └── attribut : immatriculation
```



<h3> <i class='fa fa-cogs' style="color: darkorange"> Activité 1 </h3>  
le but de cette activité est de pouvoir écrire la classe Voiture qui possédera trois attributs:

- marque
 couleur
- immatriculation  




# 🟢 GUIDÉ — La classe Voiture expliquée ligne par ligne  


<details>  
<summary> <b>Cliquez ici pour afficher la version 🟢 Niveau GUIDÉE</b>   </summary>     
    
##### Consigne   
Lis et comprend le code à l'aide des commentaires. 
```python
class Voiture:
    # __init__ est appelée automatiquement quand on crée une voiture
    def __init__(self, m, c, i):
        self.marque = m           # on stocke la marque
        self.couleur = c          # on stocke la couleur
        self.immatriculation = i  # on stocke l'immatriculation
```

## Encapsulation -  La boîte noire  
  
Imagine un distributeur de billets. Tu ne vois pas comment il fonctionne à l'intérieur — tu utilises simplement les boutons prévus.

C'est ça l'**encapsulation** : on n'accède pas directement aux attributs, on passe par des méthodes dédiées :
- **accesseur (getter)** → pour *lire* un attribut
- **mutateur (setter)** → pour *modifier* un attribut



<div class="alert alert-danger" style="color:black" markdownn="1">
   
***
<i class="fa-fw fas fa-volume-up"> **À retenir** </i>
>L’encapsulation est un des principes édictés par Alan Kay. Il consiste à considérer un objet comme une boîte noire.  
>Les attributs ne sont alors pas accessibles directement mais seulement par l’intermédiaire de méthodes dédiées:  
    - les **accesseurs** pour lire les valeurs des attributs,  
    - les **mutateurs** pour modifier les valeurs des attributs.
***  


<div class="alert alert-info" style="color:black" markdownn="1">
   
***
<i class="fa-fw fas fa-info-circle">**Remarque** </i>
>La littérature parle souvent de **getter** et de **setter** pour évoquer respectivement les accesseurs et les mutateurs.
***  

### Exemple :
```python
class Voiture:

    def __init__(self, m: str, c: str, i: str):
        self.marque = m
        self.couleur = c
        self.immatriculation = i
    
    def get_marque(self) -> str:
        """
        accesseur de la marque
        """
        return self.marque
    
    def set_marque(self, m:str) -> None:
        """
        mutateur de la marque
        """
        self.marque = m
```

On peut se poser la question de l’utilité de ces méthodes.  
En effet, en Python, il est possible d’accéder directement aux attributs.

```python
une_voiture = Voiture("Renault", "gris", "NSI-001")
# Ce code fonctionne en Python
print(une_voiture.couleur)
une_voiture.couleur = "rouge"
print(une_voiture.couleur)

```

Cependant, ce comportement n’est pas identique dans tous les langages. Le Java, par exemple, définit les attributs privés et donc inaccessibles directement.  

Surtout, l’objectif est d’obliger les utilisateurs de ne pas se détourner de l’interface établie. 

Dans l’exemple précédent, imaginons que le cahier des charges impose un changement de couleur des voitures limité à quelques propositions.
```python
 # Ce mutateur permet un contrôle sur le changement de couleur
    def set_couleur(self, c: str) -> None:
        autorisees = ["gris", "taupe", "blanc"]
        if c in autorisees:
            self.couleur = c
```

# 🟡 INTERMÉDIAIRE — Classe Voiture avec mutateur contrôlé

class Voiture:
    def __init__(self, m: str, c: str, i: str):
        self.marque = m
        self.couleur = c
        self.immatriculation = i

    def get_marque(self) -> str:
        return self.marque

    def set_couleur(self, c: str) -> None:
        couleurs_ok = ["gris", "taupe", "blanc"]
        if c in couleurs_ok:
            self.couleur = c
        # Si couleur non autorisée : rien ne se passe

# Teste différentes couleurs et observe ce qui change

# 🔴 EXPERT — Encapsulation pythonique avec @property et name mangling  

```python
class Voiture:
    def __init__(self, m: str, c: str, i: str):
        self.__marque = m           # name mangling : devient _Voiture__marque
        self.__couleur = c
        self.__immatriculation = i

    @property
    def marque(self) -> str:
        return self.__marque

    @marque.setter
    def marque(self, m: str) -> None:
        if isinstance(m, str) and m:
            self.__marque = m
        else:
            raise ValueError("La marque doit être une chaîne non vide")

# Comparaison d'accès
v = Voiture("Renault", "gris", "NSI-001")
print(v.marque)           # via @property — syntaxe identique à un attribut direct
v.marque = "Peugeot"      # via @marque.setter
print(v.marque)

# Tentative d'accès direct à l'attribut privé
try:
    print(v.__marque)     # AttributeError
except AttributeError as e:
    print(f"Erreur : {e}")

# Mais le name mangling ne protège pas totalement :
print(v._Voiture__marque) # fonctionne — donc Attention.
