```python
def rechercher_nom(carnet, nom):   # "carnet" et "nom" sont les paramètres
    for contact in carnet:
        if contact == nom:
            return True             # on renvoie le résultat
    return False
```

### 📚 Explication 2 — Spécifier une fonction : préconditions, postconditions, tests

**🎯 Objectif de ce bloc** : Écrire un jeu de tests minimal vérifiant qu'une fonction respecte sa spécification.

**🟢 Guidé** — Rappel : avant de coder, on écrit ce que la fonction attend (précondition) et ce qu'elle garantit (postcondition).
Exemple : `rechercher_nom(carnet, nom)` → précondition : `carnet` est une liste de noms ; postcondition : renvoie `True` si `nom` est dans `carnet`, `False` sinon.
Vérification : que doit renvoyer `rechercher_nom([], "Léa")` ?

**🟡 Intermédiaire** — Rappel : un jeu de tests couvre au moins un cas nominal et un cas limite (ex : liste vide, élément absent).
**Point de vigilance** : un test qui "passe" ne prouve pas que la fonction est totalement correcte, seulement qu'elle fonctionne sur ce cas précis.

**🔴 Expert** — Rappel dense : la spécification est un contrat entre l'appelant et la fonction — tant que les préconditions sont respectées, la fonction garantit ses postconditions, quelle que soit son implémentation.

### 📚 Explication 3 — Décomposer un problème en plusieurs fonctions

**🎯 Objectif de ce bloc** : Décomposer un problème en plusieurs fonctions plus simples, chacune réutilisable.

**🟢 Guidé** — Rappel : plutôt qu'un seul long programme, on écrit plusieurs petites fonctions qui s'appellent entre elles.
```python
def somme_liste(liste):
    return sum(liste)

def moyenne(liste):
    return somme_liste(liste) / len(liste)   # réutilise somme_liste
```
Vérification : pourquoi `moyenne` a-t-elle besoin de `somme_liste` plutôt que de tout recalculer elle-même ?

**🟡 Intermédiaire** — Rappel : décomposer un problème en fonctions permet de tester et corriger chaque brique séparément, avant de les combiner.
**Point de vigilance** : si `moyenne` est fausse, vérifie d'abord `somme_liste` isolément avant de chercher ailleurs.

**🔴 Expert** — Rappel dense : une décomposition réussie isole des sous-problèmes déjà rencontrés ailleurs (recherche, comptage, agrégation) pour les réutiliser sans les réécrire — c'est la compétence "reconnaître des situations déjà analysées" du programme.
