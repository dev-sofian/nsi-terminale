## 🧠 Ce que tu vas apprendre

La **Programmation Orientée Objet** (POO) est une façon de structurer le code autour de **"choses"** du monde réel (des *objets*) plutôt qu'autour de fonctions isolées.

### Les 4 concepts fondamentaux de cette partie

```
┌─────────────┐     ┌─────────────────────────────────────────┐
│   CLASSE    │ ──► │ Le "moule" ou le "plan de construction" │
│             │     │ Ex : la classe Chien                    │
└─────────────┘     └─────────────────────────────────────────┘

┌─────────────┐     ┌─────────────────────────────────────────┐
│    OBJET    │ ──► │ Une instance concrète de la classe      │
│             │     │ Ex : mon_chien = Chien("Rex", "Labrador")│
└─────────────┘     └─────────────────────────────────────────┘

┌─────────────┐     ┌─────────────────────────────────────────┐
│  ATTRIBUT   │ ──► │ Une donnée propre à l'objet             │
│             │     │ Ex : nom, race, age                     │
└─────────────┘     └─────────────────────────────────────────┘

┌─────────────┐     ┌─────────────────────────────────────────┐
│  MÉTHODE    │ ──► │ Une action que l'objet peut faire       │
│             │     │ Ex : aboyer(), manger(), courir()       │
└─────────────┘     └─────────────────────────────────────────┘
```

### Syntaxe Python de base

```python
class NomDeLaClasse:
    
    def __init__(self, parametre1, parametre2):
        """Constructeur : appelé automatiquement à la création de l'objet"""
        self.attribut1 = parametre1   # self = "moi, l'objet"
        self.attribut2 = parametre2
    
    def ma_methode(self):
        """Une méthode : une action de l'objet"""
        return f"Je suis {self.attribut1}"

# Créer un objet (= instancier la classe)
mon_objet = NomDeLaClasse("valeur1", "valeur2")
print(mon_objet.ma_methode())
```

> 💡 **`self`** est obligatoire en premier paramètre de chaque méthode. Il représente l'objet lui-même.
