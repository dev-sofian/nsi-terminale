### 📚 Explication 1 — Principe de la recherche dichotomique

**🎯 Objectif de ce bloc** : écrire un algorithme de recherche dichotomique dans une liste triée, et expliquer pourquoi elle exige un tri préalable.

**🟢 Guidé**
```python
def dichotomie(liste, valeur):
    gauche, droite = 0, len(liste) - 1
    while gauche <= droite:
        milieu = (gauche + droite) // 2
        if liste[milieu] == valeur:
            return True
        elif liste[milieu] < valeur:
            gauche = milieu + 1
        else:
            droite = milieu - 1
    return False
```
Vérification : sur `[2, 5, 8, 12, 15]`, quelle est la valeur de `milieu` au premier tour ?

**Point de vigilance** : sans tri préalable, comparer au milieu ne donne aucune information exploitable.

**🔴 Expert** — La terminaison se prouve avec un variant de boucle : `droite - gauche` diminue strictement à chaque itération.

### 📚 Explication 2 — Tri par insertion

**🎯 Objectif de ce bloc** : écrire un algorithme de tri par insertion, préalable nécessaire à l'utilisation de la dichotomie.

**🟢 Guidé**
```python
def tri_insertion(liste):
    for i in range(1, len(liste)):
        valeur = liste[i]
        j = i - 1
        while j >= 0 and liste[j] > valeur:
            liste[j + 1] = liste[j]
            j = j - 1
        liste[j + 1] = valeur
    return liste
```
Vérification : sur `[5, 2, 8]`, que contient la liste juste après la première itération (`i = 1`) ?

**🟡 Intermédiaire** — À chaque étape `i`, la partie `liste[0:i]` est déjà triée ; on y insère `liste[i]` à la bonne place en décalant les éléments plus grands vers la droite.
**Point de vigilance** : la boucle interne (`while`) doit s'arrêter dès que `liste[j] <= valeur`, sinon on décale trop d'éléments.

**🔴 Expert** — L'invariant de boucle est "au début de chaque itération `i`, `liste[0:i]` est triée".

**🎯 Objectif de ce bloc** : comparer intuitivement le coût de la recherche séquentielle, de la recherche dichotomique et du tri.

**🟢 Guidé** — Chercher dans une liste non triée oblige à tout regarder un par un ; chercher dans une liste triée permet de couper en deux à chaque fois — mais il faut d'abord trier, ce qui a aussi un coût.

**🟡 Intermédiaire** — La recherche séquentielle coûte au maximum n comparaisons ; la dichotomie coûte environ log₂(n) comparaisons, mais seulement si la liste est déjà triée.
**Point de vigilance** : trier une liste juste pour faire une seule recherche n'est pas forcément avantageux.

**🟡 Intermédiaire**
```python
def recherche_sequentielle_comptee(liste, valeur):
    comparaisons = 0
    for x in liste:
        comparaisons = comparaisons + 1
        if x == valeur:
            return comparaisons
    return comparaisons

def dichotomie_comptee(liste, valeur):
    comparaisons = 0
    gauche, droite = 0, len(liste) - 1
    while gauche <= droite:
        comparaisons = comparaisons + 1
        milieu = (gauche + droite) // 2
        if liste[milieu] == valeur:
            return comparaisons
        elif liste[milieu] < valeur:
            gauche = milieu + 1
        else:
            droite = milieu - 1
    return comparaisons

liste = list(range(0, 200, 2))   # 100 éléments triés : 0, 2, 4, ..., 198
print(recherche_sequentielle_comptee(liste, 198))
print(dichotomie_comptee(liste, 198))
```
*Consigne : exécute ce code et compare les deux nombres affichés. Recommence avec une valeur au début de la liste (`0`) : le résultat change-t-il autant pour les deux fonctions ?*

**🔴 Expert**
```python
def tri_insertion(liste):
    for i in range(1, len(liste)):
        valeur = liste[i]
        j = i - 1
        while j >= 0 and liste[j] > valeur:
            liste[j + 1] = liste[j]
            j = j - 1
        liste[j + 1] = valeur
    return liste

def dichotomie(liste, valeur):
    gauche, droite = 0, len(liste) - 1
    while gauche <= droite:
        milieu = (gauche + droite) // 2
        if liste[milieu] == valeur:
            return True
        elif liste[milieu] < valeur:
            gauche = milieu + 1
        else:
            droite = milieu - 1
    return False

def contact_existe(carnet, nom):
    carnet_trie = tri_insertion(carnet.copy())
    return dichotomie(carnet_trie, nom)
```

Pour un carnet de 15 contacts, quelle stratégie te semble la plus avantageuse ?*

### ✅ Bilan de la notion
- La recherche dichotomique divise l'intervalle de recherche par deux à chaque itération, mais exige une liste triée.
- Sa terminaison se prouve avec un variant de boucle (`droite - gauche` diminue strictement).
- Le tri par insertion place chaque élément à sa bonne place parmi les éléments déjà triés ; son invariant garantit que la partie traitée reste triée.
- La recherche séquentielle coûte au maximum n comparaisons, la dichotomie environ log₂(n) — mais le tri préalable a lui-même un coût à prendre en compte.
