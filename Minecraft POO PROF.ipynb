## 3.5 Définir les méthodes


<div class="alert alert-danger" style="color:black" markdownn="1">
                                                                  
***
<i class="fa-fw fas fa-volume-up">&nbsp;**À&nbsp;retenir**</i>
>Une **méthode** est une fonction interne à la classe. En Python, le premier paramètre d’une méthode est toujours self
***

````python
class Terre:
    def __init__(self, c: str):
        self.couleur = c
        self.est_labouree = False
        
    def labourer(self) -> None:
    """
    laboure la terre si ce n'est pas déjà fait
    """
        if not self.est_labouree:
            self.est_labouree = True
            self.couleur = "#712712"
````

<h3> <i class='fa fa-cogs' style="color: darkorange">  Activité 3: Ajout de méthode</h3>  

1. Recopier la méthode labourer(self) → None
de la classe Terre.
2. Dans la classe Roche, ajouter la méthode
**subir_impact(self, impact: int) → None**
qui applique l’impact sur la Roche.
 Si la durete devient négative ou nulle :
    - la couleur devient **#444444**,
    - la Roche est minée.

## Correction activité 3

````python
def subir_impact(self, impact: int) -> None:
    """
     applique l'impact sur la roche si elle
     n'est pas déjà minée
    Args:
    impact (int): la force du coup de pioche
    """
    if not self.est_minee:
        self.durete = self.durete - impact
        # entièrement minée
        if self.durete <= 0:
            self.couleur = "#444444"
            self.est_minee = True
````

<h3> <i class='fa fa-cogs' style="color: darkorange">  Activité 4: Ajout de méthode classe mouton </h3>    


Dans la classe Mouton, ajouter la
méthode :
1. **assommer(self) → None** qui assomme le Mouton s’il ne l’est pas,
2. **tondre(self) → int** qui tond le Mouton s’il est
assommé. La méthode devra renvoyer la quantité
de laine tondue.


````python
def assommer(self) -> None:
    """
    3 assomme le mouton
     """
    if not self.est_assomme:
        self.est_assomme = True
````

````python
def tondre(self) -> int:
     """
     tond le mouton s'il est assommé
    
     Returns:
     int: quantité de laine
     """
     quantite = 0
     if self.est_assomme:
         quantite = self.quantite_laine
         self.est_tondu = True
         # il n'a plus de laine
         self.quantite_laine = 0
         return quantite
````
