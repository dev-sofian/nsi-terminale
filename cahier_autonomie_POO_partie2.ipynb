## Pourquoi encapsuler ?

Sans encapsulation, n'importe qui peut modifier les attributs d'un objet directement, même avec des valeurs aberrantes :

```python
ma_voiture.carburant = -999   # 😱 personne ne peut l'en empêcher
ma_voiture.age = "cinq"       # 😱 type incorrect
```

L'**encapsulation** consiste à **protéger les attributs** en les rendant privés, et à contrôler leur accès via des méthodes dédiées.

## Conventions Python

```python
class MaClasse:
    def __init__(self):
        self.public       = "accessible partout"
        self._protege     = "convention : ne pas toucher de l'extérieur"
        self.__prive      = "vraiment masqué (name mangling)"
    
    # Getter : lire la valeur
    def get_prive(self):
        return self.__prive
    
    # Setter : écrire la valeur avec contrôle
    def set_prive(self, valeur):
        if not isinstance(valeur, str):
            raise TypeError("Doit être une chaîne")
        self.__prive = valeur
    
    # Version Pythonique avec @property
    @property
    def prive(self):
        return self.__prive
    
    @prive.setter
    def prive(self, valeur):
        if not isinstance(valeur, str):
            raise TypeError("Doit être une chaîne")
        self.__prive = valeur
```

```python
obj = MaClasse()
print(obj.prive)       # appelle le getter automatiquement
obj.prive = "nouveau"  # appelle le setter automatiquement
```

> 💡 **Lien Pacman :** Le score de Pacman ne doit jamais être négatif. L'énergie ne doit pas dépasser 100. L'encapsulation permettra de garantir ces invariants.

## Le principe

L'héritage permet de créer une nouvelle classe en **réutilisant** le code d'une classe existante, en ajoutant ou modifiant des comportements.

```
        Animal
       /      \
    Chien     Chat
    /   \
Labrador  Berger
```

```python
class Animal:
    def __init__(self, nom, age):
        self.nom = nom
        self.age = age
    
    def se_presenter(self):
        return f"Je m'appelle {self.nom}, j'ai {self.age} an(s)."
    
    def faire_son(self):
        return "..."    # méthode à redéfinir dans les sous-classes


class Chien(Animal):              # Chien HÉRITE de Animal
    def __init__(self, nom, age, race):
        super().__init__(nom, age)  # appelle le __init__ de Animal
        self.race = race            # attribut spécifique à Chien
    
    def faire_son(self):           # REDÉFINIT la méthode de Animal
        return f"{self.nom} dit : Woof !"
    
    def rapporter(self, objet):    # méthode PROPRE à Chien
        return f"{self.nom} rapporte {objet} !"
```

> 💡 **`super()`** appelle la méthode de la classe mère. Toujours l'utiliser dans `__init__` d'une classe fille pour ne pas perdre l'initialisation de la mère.

> 💡 **Lien Pacman :** `Ennemi` sera la classe mère. `Fantome`, `SuperFantome`, `Boss` seront des classes filles avec des comportements différents.

## Le principe

Le **polymorphisme** signifie qu'un même appel de méthode produit des comportements **différents selon l'objet**. C'est la conséquence naturelle de l'héritage + la redéfinition.

```python
animaux = [Chien("Rex", 3, "Labrador"), Chaton("Félix", 2, "gris"), Animal("Tweety", 1)]

for animal in animaux:
    print(animal.faire_son())   # même appel, comportement différent !
# Rex dit : Woof !
# Félix dit : Miaou !
# ...
```

**Duck Typing** (Python) : "Si ça marche comme un canard et cancanine comme un canard, c'est un canard."
→ Python vérifie le comportement, pas le type. Si un objet a la méthode `faire_son()`, on peut l'appeler, peu importe sa classe.

## Méthodes magiques (dunder methods)

```python
class Vecteur:
    def __init__(self, x, y):
        self.x, self.y = x, y
    
    def __str__(self):  return f"({self.x}, {self.y})"
    def __repr__(self): return f"Vecteur({self.x}, {self.y})"
    def __add__(self, other): return Vecteur(self.x + other.x, self.y + other.y)
    def __mul__(self, scalaire): return Vecteur(self.x * scalaire, self.y * scalaire)
    def __eq__(self, other):  return self.x == other.x and self.y == other.y
    def __len__(self): return int((self.x**2 + self.y**2)**0.5)

v1 = Vecteur(1, 2)
v2 = Vecteur(3, 4)
print(v1 + v2)     # appelle __add__  → (4, 6)
print(v1 * 3)      # appelle __mul__  → (3, 6)
print(v1 == v2)    # appelle __eq__   → False
```

> 💡 **Lien Pacman :** `se_deplacer()` sera polymorphe — Pacman, Blinky, Pinky, Inky, Clyde bougent tous différemment mais avec le même appel.

## Composition vs Héritage

| | Héritage | Composition |
|---|----------|-------------|

| Exemple | `Chien` **est un** `Animal` | `Voiture` **a un** `Moteur` |
| Pacman | `Fantôme` **est un** `Ennemi` | `Jeu` **a une** `Grille` et des `Fantômes` |

    def __init__(self, puissance):
        self.puissance = puissance

    def demarrer(self): self.en_marche = True
    def arreter(self):  self.en_marche = False


class Voiture:
    def __init__(self, marque, puissance):
        self.marque  = marque
        self.moteur  = Moteur(puissance)   # composition : Voiture CONTIENT un Moteur
    
    def demarrer(self):
        self.moteur.demarrer()             # délégation : on délègue au moteur
        print(f"{self.marque} démarre !")
