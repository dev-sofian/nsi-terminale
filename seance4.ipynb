### 📚 Explication 3 — Recherche séquentielle : occurrence, extremum, moyenne

**🎯 Objectif de ce bloc** : écrire un algorithme de recherche séquentielle et comprendre pourquoi son coût est linéaire.

**🟢 Guidé**
```python
def rechercher_nom(carnet, nom):
    for contact in carnet:
        if contact == nom:
            return True
    return False
```
Vérification : si `nom` apparaît deux fois, combien de fois la boucle continue-t-elle après l'avoir trouvé une première fois ?

**🟡 Intermédiaire** — Un arrêt anticipé (`return` dès que trouvé) évite un travail inutile.
**Point de vigilance** : compter les occurrences nécessite au contraire de *ne pas* s'arrêter à la première trouvée.

**🔴 Expert** — Dans le pire cas, le nombre de comparaisons d'une recherche séquentielle est proportionnel à la taille de la liste.

**🔴 Expert**
```python
def rechercher_nom_indexe(carnet, nom):
    for i in range(len(carnet)):
        if carnet[i] == nom:
            return (True, i)
    return (False, -1)

carnet = ["Léa", "Tom", "Zoé"]
print(rechercher_nom_indexe(carnet, "Zoé"))   # attendu : (True, 2)
```

### ✅ Bilan de la notion
- Parcourir une liste se fait par valeur (`for x in liste`) ou par index (`for i in range(len(liste))`) selon le besoin de modification.
- Une compréhension `[expression for x in liste if condition]` construit une nouvelle liste en une ligne.
- Une recherche séquentielle compare un par un ; un arrêt anticipé évite un travail inutile quand on cherche une seule occurrence.
- Le coût d'une recherche séquentielle est linéaire : proportionnel à la taille de la liste dans le pire cas.
